# 02 - Data Cleaning

```python
import pandas as pd

df = pd.read_csv('../data/raw/food_delivery.csv')
df.shape
```

```python
df.isnull().sum()
```

Drop `Order_ID`, it's just a row identifier and won't help the model.

```python
df = df.drop(columns=['Order_ID'])
```

```python
df.duplicated().sum()
```

No duplicates. 
For the missing values:
- categorical columns (`Weather`, `Traffic_Level`, `Time_of_Day`) -> fill with the most common value
- `Courier_Experience_yrs` -> fill with the median

```python
cat_cols = ['Weather', 'Traffic_Level', 'Time_of_Day']

for col in cat_cols:
    mode_val = df[col].mode()[0]
    df[col] = df[col].fillna(mode_val)

df['Courier_Experience_yrs'] = df['Courier_Experience_yrs'].fillna(df['Courier_Experience_yrs'].median())
```

```python
df.isnull().sum()
```

Quick sanity check on ranges before saving.

```python
df[['Distance_km', 'Preparation_Time_min', 'Courier_Experience_yrs', 'Delivery_Time_min']].describe()
```
